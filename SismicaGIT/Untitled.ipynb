```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("10002_fb.txt")
df
```

```python
plt.scatter(df[' X'],df[' Time'])
plt.xlabel('Posición')
plt.ylabel('Tiempo')
```

```python
m, b = np.polyfit(df[' X'].to_list()[:6], df[' Time'].to_list()[:6], 1)

print("Velocidad:", 1/m)
print("Intercept (b):", b)
```

```python
m, b = np.polyfit(df[' X'].to_list()[6:], df[' Time'].to_list()[6:], 1)

print("Velocidad:", 1/m)
print("Intercept (b):", b)
```
