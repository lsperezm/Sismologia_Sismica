```python
import math
import numpy as np
import matplotlib.pyplot as plt
from obspy import read
import matplotlib.dates as mdates


from obspy.taup import TauPyModel
```

```python
Tau=read('2016-11-13mww78-south-island-new-zealand-1.miniseed')
SNZO=read('2016-11-13mww78-south-island-new-zealand-2.miniseed')
WRAB=read('2016-11-13mww78-south-island-new-zealand.miniseed')
Tau
```

```python

target_timeP3 = Tau[0].stats.starttime + 1*60 + 0  
target_mpl_timeP3 = mdates.date2num(target_timeP3.datetime)

Tau[0].stats.starttime , target_mpl_timeP3
```

```python
fig, ax = plt.subplots(3, 1, sharex=True, figsize=(10, 7))
#ax[0].plot(file[0].times("matplotlib"), file[0].data, color='gray',label='BH1')
ax[0].plot(Tau[0].times("matplotlib"), Tau[0].data, color='black',label='TAUBH1')

#ax[1].plot(file[1].times("matplotlib"), file[1].data, color='gray',label='BH2')
ax[1].plot(Tau[1].times("matplotlib"), Tau[1].data, color='black',label='TAUBH2')

#ax[2].plot(file[2].times("matplotlib"), file[2].data, color='gray',label='BHZ')
ax[2].plot(Tau[2].times("matplotlib"), Tau[2].data, color='black',label='TAUBHZ')

ax[2].xaxis_date()
for axn in ax:
    axn.set_ylabel('Amplitud') 
    axn.axvline(target_mpl_timeP3+0.00001, color='mediumblue', linestyle='--', linewidth=1.5, label='Arribo P')

    axn.legend(loc='upper left')

plt.xlim(Tau[0].times("matplotlib")[0],Tau[0].times("matplotlib")[4520])
```
